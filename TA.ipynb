```python
# prompt: use deepl to translate dataframe named "embarrassment_data.csv" to bahasa indonesia

import pandas as pd
import deepl
import time

# Assuming you have your DeepL API key set as an environment variable
# or replace with your actual key
try:
    translator = deepl.Translator("41743afb-058e-4c43-adf3-abbd6abd3073:fx")
except deepl.exceptions.DeepLException as e:
    print(f"DeepL API error: {e}")
    print("Please ensure your API key is correct and set.")
    exit()

try:
    df = pd.read_csv('embarrassment_data.csv')
except FileNotFoundError:
    print("Error: 'embarrassment_data.csv' not found.")
    exit()

df_translated = pd.DataFrame(columns=['text', 'label'])

# Process all lines in the DataFrame
for i in range(len(df)):
    try:
        original_text = df['text'][i]
        result = translator.translate_text(original_text, target_lang="ID")
        translated_text = result.text
        df_translated = pd.concat([df_translated, pd.DataFrame({'text': [translated_text], 'label': [df['label'][i]]})], ignore_index=True)
        # Add a delay to prevent hitting the API rate limit (adjust as needed)
        time.sleep(0.1)
        # Print progress
        if i % 100 == 0:
            print(f"Processed {i} lines...")
    except Exception as e:
        print(f"Error translating line {i}: {e}")
        # Handle the error, skip the line or log it appropriately.
        continue

print("Translation complete.")

# Save the translated DataFrame to a new CSV file
df_translated.to_csv('embarrassment_data_id.csv', index=False)
print("Translated data saved to 'embarrassment_data_id.csv'")
```

```python
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from torch.utils.data import Dataset, DataLoader
from transformers import BertTokenizer, BertForSequenceClassification, Trainer, TrainingArguments
import ast

# 1. Load Dataset
dataset_path = "/content/drive/MyDrive/Tugas Akhir/cleaned_dataset_with_stopwords.csv"
print("📥 Loading dataset...")
df = pd.read_csv(dataset_path)

# 2. Convert tokenized text back into readable sentences
print("🔄 Converting tokenized text back to sentences...")
df['Teks'] = df['Teks'].apply(ast.literal_eval)  # Convert string list to actual list
df['Teks'] = df['Teks'].apply(lambda x: ' '.join(x))  # Join tokens into a single string

# 3. Extract text and labels
print("📌 Extracting text and labels...")
texts = df['Teks'].tolist()
labels = df[['Senang', 'Sedih', 'Marah', 'Iri', 'Malu', 'Takut']].values

# 4. Split dataset into training and validation sets
print("✂️ Splitting dataset into training (80%) and validation (20%)...")
train_texts, val_texts, train_labels, val_labels = train_test_split(
    texts, labels, test_size=0.2, random_state=42
)

# 5. Load IndoBERT tokenizer
print("🔠 Loading IndoBERT tokenizer...")
tokenizer = BertTokenizer.from_pretrained("indobenchmark/indobert-base-p1")

# 6. Tokenization function
def tokenize_function(texts):
    return tokenizer(texts, padding=True, truncation=True, max_length=512, return_tensors='pt')

# 7. Define Dataset Class
class EmotionDataset(Dataset):
    def __init__(self, texts, labels):
        self.encodings = tokenize_function(texts)
        self.labels = torch.tensor(labels, dtype=torch.float)  # Multi-label classification

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        item = {key: val[idx] for key, val in self.encodings.items()}
        item['labels'] = self.labels[idx]
        return item

# 8. Create dataset objects
print("📦 Creating PyTorch datasets...")
train_dataset = EmotionDataset(train_texts, train_labels)
val_dataset = EmotionDataset(val_texts, val_labels)

# 9. Load IndoBERT model for multi-label classification
print("🧠 Loading IndoBERT model...")
model = BertForSequenceClassification.from_pretrained(
    "indobenchmark/indobert-base-p1", num_labels=6, problem_type="multi_label_classification"
)

# 10. Define Training Arguments
print("⚙️ Setting up training arguments...")
training_args = TrainingArguments(
    output_dir="./results",
    evaluation_strategy="epoch",
    save_strategy="epoch",
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    num_train_epochs=3,
    weight_decay=0.01,
    logging_dir="./logs",
    logging_steps=10,
    save_total_limit=2  # Keep only last 2 checkpoints
)

# 11. Initialize Trainer
print("🚀 Initializing Trainer...")
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=val_dataset,
)

# 12. Train the Model
print("🎯 Starting training...")
trainer.train()

# 13. Save the Model and Tokenizer
print("💾 Saving model and tokenizer...")
model.save_pretrained("./indobert_emotion_model")
tokenizer.save_pretrained("./indobert_emotion_model")

print("✅ Training complete! Model saved in './indobert_emotion_model'")
```

```python
# prompt: check accuracy etc of /content/indobert_emotion_model

from sklearn.metrics import accuracy_score, precision_recall_fscore_support

# Assuming you have your trained model and tokenizer loaded
model = BertForSequenceClassification.from_pretrained("/content/indobert_emotion_model")
tokenizer = BertTokenizer.from_pretrained("/content/indobert_emotion_model")

# Load your validation dataset (replace with your actual dataset)
val_dataset = EmotionDataset(val_texts, val_labels)

# Function to predict labels for a given dataset
def predict_labels(dataset):
    predictions = []
    for batch in DataLoader(dataset, batch_size=8):
        inputs = {k: v.to(model.device) for k, v in batch.items() if k != 'labels'}
        outputs = model(**inputs)
        predicted_labels = (torch.sigmoid(outputs.logits) > 0.5).cpu().numpy().astype(int)
        predictions.extend(predicted_labels)
    return predictions

# Make predictions on the validation set
predicted_labels = predict_labels(val_dataset)

# Calculate accuracy, precision, recall, F1-score
accuracy = accuracy_score(val_labels, predicted_labels)
precision, recall, f1_score, _ = precision_recall_fscore_support(val_labels, predicted_labels, average='weighted')

print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-Score: {f1_score:.4f}")

# You can also calculate per-class metrics if needed
# precision, recall, f1_score, _ = precision_recall_fscore_support(val_labels, predicted_labels, average=None)
# print("Per-class metrics:")
# for i in range(6):  # Assuming 6 emotion labels
#     print(f"Label {i}: Precision={precision[i]:.4f}, Recall={recall[i]:.4f}, F1-Score={f1_score[i]:.4f}")
```
